```python

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
from pathlib import Path

from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import SimpleImputer, KNNImputer, IterativeImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split, RepeatedStratifiedKFold, cross_validate
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

# ============================================================
# STEP 1: LOAD ORIGINAL DATASET
# ============================================================

df = pd.read_csv("../chronic_dataset/chronic_kidney_disease_corrected.csv")

numerical_columns = [
    "age", "bp", "sg", "al", "su", "bgr", "bu",
    "sc", "sod", "pot", "hemo", "pcv", "wbcc", "rbcc"
]

categorical_columns = [
    "rbc", "pc", "pcc", "ba", "htn",
    "dm", "cad", "appet", "pe", "ane"
]

for col in numerical_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

for col in categorical_columns:
    df[col] = (
        df[col].astype(str).str.strip().str.lower()
        .replace(["?", "", "nan", "none", "<na>"], np.nan)
    )

df["class"] = (
    df["class"].astype(str).str.strip().str.lower()
    .map({"ckd": 1, "notckd": 0})
)

if df["class"].isna().any():
    raise ValueError("Invalid target labels detected.")

X = df[numerical_columns + categorical_columns]
y = df["class"].astype(int)

# ============================================================
# STEP 2: OUTER TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)

# ============================================================
# STEP 3: CREATE IMPUTATION PIPELINES
# ============================================================

def create_preprocessor(method):

    if method == "Median-Mode":
        numerical_pipeline = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler())
        ])

    elif method == "KNN":
        numerical_pipeline = Pipeline([
            ("scaler_before", StandardScaler()),
            ("imputer", KNNImputer(n_neighbors=5)),
            ("scaler_after", StandardScaler())
        ])

    elif method == "Iterative":
        numerical_pipeline = Pipeline([
            ("scaler_before", StandardScaler()),
            ("imputer", IterativeImputer(
                max_iter=30,
                random_state=42
            )),
            ("scaler_after", StandardScaler())
        ])

    categorical_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        ))
    ])

    return ColumnTransformer([
        ("num", numerical_pipeline, numerical_columns),
        ("cat", categorical_pipeline, categorical_columns)
    ])

# ============================================================
# STEP 4: DEFINE MODELS
# ============================================================

def create_models():
    return {
        "Logistic Regression": LogisticRegression(
            max_iter=2000, random_state=42
        ),
        "Random Forest": RandomForestClassifier(
            n_estimators=200,
            random_state=42,
            n_jobs=1
        ),
        "SVM": SVC(
            kernel="rbf",
            probability=True,
            random_state=42
        )
    }

# ============================================================
# STEP 5: REPEATED STRATIFIED CROSS-VALIDATION
# ============================================================

cv = RepeatedStratifiedKFold(
    n_splits=5,
    n_repeats=10,
    random_state=42
)

scoring = {
    "Accuracy": "accuracy",
    "Precision": "precision",
    "CKD_Recall": "recall",
    "Specificity": "recall",
    "F1_Score": "f1",
    "ROC_AUC": "roc_auc"
}

all_results = []

for method in ["Median-Mode", "KNN", "Iterative"]:

    print("\n" + "=" * 65)
    print("IMPUTATION METHOD:", method)
    print("=" * 65)

    for model_name, model in create_models().items():

        pipeline = Pipeline([
            ("preprocessor", create_preprocessor(method)),
            ("classifier", model)
        ])

        scores = cross_validate(
            pipeline,
            X_train,
            y_train,
            cv=cv,
            scoring=scoring,
            n_jobs=1,
            error_score="raise"
        )

        result = {
            "Imputation": method,
            "Model": model_name,
            "Accuracy_Mean": scores["test_Accuracy"].mean(),
            "Accuracy_STD": scores["test_Accuracy"].std(),
            "Precision_Mean": scores["test_Precision"].mean(),
            "CKD_Recall_Mean": scores["test_CKD_Recall"].mean(),
            "Specificity_Mean": scores["test_Specificity"].mean(),
            "F1_Mean": scores["test_F1_Score"].mean(),
            "ROC_AUC_Mean": scores["test_ROC_AUC"].mean()
        }

        all_results.append(result)

        print(
            f"{model_name}: "
            f"Accuracy={result['Accuracy_Mean']:.4f}, "
            f"F1={result['F1_Mean']:.4f}, "
            f"ROC-AUC={result['ROC_AUC_Mean']:.4f}"
        )

# ============================================================
# STEP 6: ALL MODEL RESULTS
# ============================================================

results_df = pd.DataFrame(all_results)

results_df = results_df.sort_values(
    ["F1_Mean", "ROC_AUC_Mean", "Accuracy_Mean"],
    ascending=False
).reset_index(drop=True)

print("\nALL 9 MODEL RESULTS")
print(results_df.round(4).to_string(index=False))

# ============================================================
# STEP 7: IMPUTATION METHOD COMPARISON
# ============================================================

comparison = results_df.groupby("Imputation")[
    [
        "Accuracy_Mean",
        "Precision_Mean",
        "CKD_Recall_Mean",
        "Specificity_Mean",
        "F1_Mean",
        "ROC_AUC_Mean"
    ]
].mean().reset_index()

comparison = comparison.sort_values(
    ["F1_Mean", "ROC_AUC_Mean", "Accuracy_Mean"],
    ascending=False
).reset_index(drop=True)

print("\nIMPUTATION METHOD COMPARISON")
print(comparison.round(4).to_string(index=False))

# ============================================================
# STEP 8: SELECT BEST METHOD AND MODEL
# ============================================================

best = results_df.iloc[0]

best_method = best["Imputation"]
best_model_name = best["Model"]

print("\nBEST IMPUTATION METHOD:", best_method)
print("BEST MODEL:", best_model_name)
print("CV ACCURACY:", round(best["Accuracy_Mean"], 4))
print("CV F1-SCORE:", round(best["F1_Mean"], 4))
print("CV ROC-AUC:", round(best["ROC_AUC_Mean"], 4))

# ============================================================
# STEP 9: TRAIN BEST MODEL ON COMPLETE TRAINING DATA
# ============================================================

best_pipeline = Pipeline([
    ("preprocessor", create_preprocessor(best_method)),
    ("classifier", create_models()[best_model_name])
])

best_pipeline.fit(X_train, y_train)

# ============================================================
# STEP 10: FINAL TEST EVALUATION
# ============================================================

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix,
    classification_report
)

y_pred = best_pipeline.predict(X_test)
y_prob = best_pipeline.predict_proba(X_test)[:, 1]

test_results = {
    "Accuracy": accuracy_score(y_test, y_pred),
    "Precision": precision_score(y_test, y_pred, zero_division=0),
    "CKD_Recall": recall_score(y_test, y_pred, zero_division=0),
    "Specificity": recall_score(y_test, y_pred, pos_label=0, zero_division=0),
    "F1_Score": f1_score(y_test, y_pred, zero_division=0),
    "ROC_AUC": roc_auc_score(y_test, y_prob)
}

print("\nFINAL TEST RESULTS")
for metric, value in test_results.items():
    print(f"{metric}: {value:.4f}")

print("\nCONFUSION MATRIX")
print(confusion_matrix(y_test, y_pred, labels=[1, 0]))

print("\nCLASSIFICATION REPORT")
print(classification_report(
    y_test,
    y_pred,
    labels=[0, 1],
    target_names=["Not CKD", "CKD"],
    zero_division=0
))

# ============================================================
# STEP 11: SAVE RESULTS
# ============================================================

output_folder = Path("../model_results")
output_folder.mkdir(parents=True, exist_ok=True)

results_df.to_csv(
    output_folder / "leakage_free_repeated_cv_all_models.csv",
    index=False
)

comparison.to_csv(
    output_folder / "leakage_free_imputation_comparison.csv",
    index=False
)

pd.DataFrame([{
    "Best_Imputation": best_method,
    "Best_Model": best_model_name,
    **test_results
}]).to_csv(
    output_folder / "best_model_final_test_results.csv",
    index=False
)

print("\nAll results saved successfully.")
```

```python
# STEP 1: SAVE THE COMPLETE TRAINED MODEL PIPELINE

import os
import joblib
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.svm import SVC

# 1. Load dataset
data_path = "../chronic_dataset/chronic_kidney_disease_corrected.csv"
df = pd.read_csv(data_path)

# 2. Separate features and target
df.columns = df.columns.str.strip()

numerical_columns = [
    "age", "bp", "sg", "al", "su", "bgr", "bu",
    "sc", "sod", "pot", "hemo", "pcv", "wbcc", "rbcc"
]

categorical_columns = [
    "rbc", "pc", "pcc", "ba", "htn",
    "dm", "cad", "appet", "pe", "ane"
]

X = df.drop(columns=["class"]).copy()

y = (
    df["class"]
    .astype(str)
    .str.strip()
    .str.lower()
    .map({"ckd": 1, "notckd": 0})
)

# 3. Clean feature values
for column in numerical_columns:
    X[column] = pd.to_numeric(X[column], errors="coerce")

for column in categorical_columns:
    X[column] = (
        X[column]
        .astype(str)
        .str.strip()
        .str.lower()
        .replace(["?", "", "nan", "none", "<na>"], float("nan"))
    )

# 4. Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# 5. Numerical preprocessing
numerical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# 6. Categorical preprocessing
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

# 7. Combine preprocessing
preprocessor = ColumnTransformer([
    ("numerical", numerical_pipeline, numerical_columns),
    ("categorical", categorical_pipeline, categorical_columns)
])

# 8. Create complete model pipeline
ckd_model_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", SVC(
        kernel="rbf",
        probability=True,
        random_state=42
    ))
])

# 9. Train on training data
ckd_model_pipeline.fit(X_train, y_train)

# 10. Save complete pipeline
os.makedirs("../models", exist_ok=True)

model_path = "../models/ckd_median_mode_svm_pipeline.joblib"

joblib.dump(ckd_model_pipeline, model_path)

# 11. Verify saved model
loaded_model = joblib.load(model_path)

print("Model Pipeline Saved Successfully!")
print("Model Path:", os.path.abspath(model_path))
print("Training Records:", len(X_train))
print("Testing Records:", len(X_test))
print("Model:", loaded_model.named_steps["classifier"])
print("Preprocessing and Model:", list(loaded_model.named_steps.keys()))
```

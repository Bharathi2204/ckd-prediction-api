## median mode 

```python

import pandas as pd
from pathlib import Path
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

# STEP 1: Load Cleaned Training and Testing Datasets
train_data = pd.read_csv("../cleaned_datasets/median_mode_train_clean.csv")
test_data = pd.read_csv("../cleaned_datasets/median_mode_test_clean.csv")

X_train = train_data.drop(columns=["class"])
y_train = train_data["class"].map({"ckd": 0, "notckd": 1})

X_test = test_data.drop(columns=["class"])
y_test = test_data["class"].map({"ckd": 0, "notckd": 1})

# STEP 2: Identify Numerical and Categorical Columns
numerical_columns = X_train.select_dtypes(include=["number"]).columns.tolist()
categorical_columns = X_train.select_dtypes(include=["object", "str"]).columns.tolist()

print("Numerical Columns:", len(numerical_columns))
print("Categorical Columns:", len(categorical_columns))

# STEP 3: Encode Categorical Features and Scale Numerical Features
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_columns),
        ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False),
         categorical_columns)
    ]
)

# Fit only on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform testing data
X_test_processed = preprocessor.transform(X_test)

# STEP 4: Convert Processed Data into DataFrames
feature_names = preprocessor.get_feature_names_out()

X_train_processed = pd.DataFrame(
    X_train_processed,
    columns=feature_names,
    index=X_train.index
)

X_test_processed = pd.DataFrame(
    X_test_processed,
    columns=feature_names,
    index=X_test.index
)

# STEP 5: Add Target Column
X_train_processed["class"] = y_train.values
X_test_processed["class"] = y_test.values

# STEP 6: Save Preprocessed Datasets
output_folder = Path("../preprocessed_datasets")
output_folder.mkdir(parents=True, exist_ok=True)

X_train_processed.to_csv(
    output_folder / "median_mode_train.csv", index=False
)

X_test_processed.to_csv(
    output_folder / "median_mode_test.csv", index=False
)

# STEP 7: Display Results
print("\nTraining Dataset Shape:", X_train_processed.shape)
print("Testing Dataset Shape:", X_test_processed.shape)

print("Training Missing Values:", X_train_processed.isnull().sum().sum())
print("Testing Missing Values:", X_test_processed.isnull().sum().sum())

print("\nTraining Class Distribution:")
print(X_train_processed["class"].value_counts())

print("\nTesting Class Distribution:")
print(X_test_processed["class"].value_counts())

print("\nMedian-Mode preprocessing completed successfully.")
```
